```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=4,test_size=0.2)
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
reg=SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
reg.fit(X_train,y_train)
y_pred=reg.predict(X_test)
print("R2 Score:",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

```python
import numpy as np

class RidgeGD:
    def __init__(self, learning_rate, alpha, epochs) -> None:
        self.learning_rate = learning_rate
        self.alpha = alpha
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0

        theta = np.insert(self.coef_, 0, self.intercept_)
        X_train = np.insert(X_train, 0, 1, axis=1)

        n = X_train.shape[0]

        for i in range(self.epochs):
            y_pred = np.dot(X_train, theta)


            theta_der = (np.dot(X_train.T, (y_pred - y_train)) / n) + self.alpha * theta


            theta_der[0] = theta_der[0] + self.alpha * theta[0]

            theta = theta - self.learning_rate * theta_der

        self.coef_ = theta[1:]
        self.intercept_ = theta[0]

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
reg=RidgeGD(epochs=500,alpha=0.001,learning_rate=0.005)
```

```python
reg.fit(X_train,y_train)
y_pred=reg.predict(X_test)

print("R2:",r2_score(y_test,y_pred))
```
