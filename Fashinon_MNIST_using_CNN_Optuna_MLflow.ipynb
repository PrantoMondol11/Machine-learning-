```python
from torch.utils.data import Dataset,DataLoader
```

```python
class CustomDataset(Dataset):
    def __init__(self,csv_file):
        self.df=pd.read_csv(csv_file)

    def __len__(self):
        return len(self.df)

    def __getitem__(self,idx):
        X_train=self.df.iloc[:,1:].to_numpy()
        y_train=self.df.iloc[:,0].to_numpy()

        return torch.tensor(X_train[idx],dtype=torch.float32),torch.tensor(y_train[idx],dtype=torch.long)
```

```python
dataset=CustomDataset(csv_file)
csv_file2="/kaggle/input/datasets/organizations/zalando-research/fashionmnist/fashion-mnist_test.csv"
dataset2=CustomDataset(csv_file2)
```

```python
import torch.nn as nn
import torch
import optuna
import mlflow.pytorch
import mlflow
class Mymodel(nn.Module):
    def __init__(self,hidden_dim,output_dim,kernel_size,filter1,filter2):
        super().__init__()
        self.features=nn.Sequential(
            nn.LazyConv2d(filter1,kernel_size=kernel_size,padding='same'),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2),
            nn.Conv2d(filter1,filter2,kernel_size=kernel_size,padding='same'),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2)
        )
        self.network=nn.Sequential(
            nn.LazyLinear(hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim,output_dim)
        )

    def forward(self,x):
        output=self.features(x)
        output=torch.flatten(output,start_dim=1)
        output=self.network(output)

        return output



def objective(trail):

    hidden_dim=trail.suggest_categorical("hidden_dim",[16,32,64,128])
    filter1=trail.suggest_categorical("filter1",[32,64,128])
    filter2=trail.suggest_categorical("filter2",[32,64,128])
    lr=trail.suggest_float("lr",1e-5,1e-1,log=True)
    kernel_size=trail.suggest_categorical("kernel_size",[2,3])
    loss_func=trail.suggest_categorical("loss_func",["cross_entropy","label_smoothing"])
    optimizer=trail.suggest_categorical("optimizer",["Adam","SGD", "RMSprop"])
    batch_size=trail.suggest_categorical("batch_size",[16,32,64])

    with mlflow.start_run(nested=True):
        mlflow.log_params({
            "hidden_dim": hidden_dim,
            "filter1":filter1,
            "filter2":filter2,
            "learning_rate":lr,
            "kernel_size":kernel_size,
            "loss_func":loss_func,
            "optimizer":optimizer,
            "batch_size":batch_size,
            "trail":trail.number
        })


        dataloader=DataLoader(dataset,batch_size=batch_size,shuffle=True)
        tes_loader=DataLoader(dataset2,batch_size=batch_size,shuffle=False)



        model=Mymodel(hidden_dim,10,kernel_size,filter1,filter2).to(device)
        if loss_func=="cross_entropy":
            loss_fn=nn.CrossEntropyLoss()
        elif loss_func=="label_smoothing":
            loss_fn=nn.CrossEntropyLoss(label_smoothing=0.1)

        if optimizer=="Adam":
            optimizer=torch.optim.Adam(model.parameters(),lr=lr)
        elif optimizer=="SGD":
            optimizer=torch.optim.SGD(model.parameters(),lr=lr)
        else:
            optimizer=torch.optim.RMSprop(model.parameters(),lr=lr)


        epochs=10

        for epoch in range(epochs):
            total_loss=0
            for batch_features, batch_label in dataloader:
                batch_features=batch_features.reshape(-1, 1, 28, 28)
                batch_features, batch_label=batch_features.to(device),batch_label.to(device)
                output= model(batch_features)
                loss=loss_fn(output,batch_label)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                total_loss+=loss.item()
            avg_loss=total_loss/len(dataloader)

    # print(f"epoch: { epoch+1} , Loss: {avg_loss}")
    # #

        mlflow.log_metric(
                    "train_loss",
                    avg_loss,
                    step=epoch
                )
        model.eval()
        with torch.no_grad():
            total=0
            correct=0
            for batch_features,batch_label in tes_loader:
                batch_feature=batch_features.reshape(-1,1,28,28)
                batch_feature, batch_label=batch_feature.to(device),batch_label.to(device)
                output=model(batch_feature)
                predicted=output.argmax(dim=1)
                total+=batch_label.size(0)
                correct+=(batch_label==predicted).sum().item()
            accuracy = correct / total

        mlflow.log_metric(
                "validation_accuracy",
                    accuracy
                    )
        return accuracy

```

```python
import dagshub
dagshub.init(repo_owner='mondolpranto83', repo_name='Machine-learning-', mlflow=True)
mlflow.set_experiment("CNN-Optuna")
study=optuna.create_study(direction="maximize")
study.optimize(objective,n_trials=10)
print("Best accuracy:", study.best_value)
print("Best parameters:", study.best_params)
```
