```python
import pandas as pd
df=pd.read_csv('/kaggle/input/datasets/safaruzzamanshovo/graduate-admission-dataset/graduate_admission1.csv')
```

```python
df.head()
```

```python
# features and target
X = df.drop('Chance of Admit', axis=1)
y = df['Chance of Admit']
```

```python
from sklearn.model_selection import train_test_split
X_train,y_train,X_test,y_test=train_test_split(X,y,random_state=1,test_size=.2)
```

```python
X_train
X_test = X_test.to_frame()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(7,activation='relu',input_dim=7))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.compile(loss='mse',optimizer='Adam')
```

```python
history=model.fit(X_train_scaled,y_train,epochs=10,validation_split=.2)
```
