```python
X_train_tensor=X_train_tensor/255.00
X_test_tensor=X_test_tensor/255.00
```

```python
from torch.utils.data import Dataset,DataLoader
class CustomDataset(Dataset):
    def __init__(self,features,label):
        self.features=features
        self.label=label
    def __len__(self):
        return len(self.features)

    def __getitem__(self,idx):
        x=self.features[idx]
        y=self.label[idx]
        return x,y
```

```python
dataset=CustomDataset(X_train_tensor,y_train_tensor)
dataset2=CustomDataset(X_test_tensor,y_test_tensor)
train_loader=DataLoader(dataset2,batch_size=32,shuffle=False)
dataloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
import torch.nn as nn
```

```python
class Model1(nn.Module):
    def __init__(self,num_features):
        super().__init__()
        self.model=nn.Sequential(
            nn.Linear(num_features,128),
            nn.ReLU(),
            nn.Linear(128,64),
            nn.ReLU(),
            nn.Linear(64,10)
        )
    def forward(self,x):
        return self.model(x)
```

```python
loss1=nn.CrossEntropyLoss()
model=Model1(X_train_tensor.shape[1])
```

```python
optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
```

```python
epochs=100
```

```python
for epoch in range(epochs):
    total_epoch_loss=0
    for batch_features,batch_labels in dataloader:
        output= model(batch_features)
        loss=loss1(output,batch_labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_epoch_loss+=loss.item()

    avg_loss=total_epoch_loss/len(dataloader)
    print(f"Epoch: {epoch+1} , Avg Loss {avg_loss}")


```

```python
model.eval()
```

```python
total=0
correct=0

with torch.no_grad():
    for batch_features,batch_labels in train_loader:
        outputs=model(batch_features)
        _,predictes=torch.max(outputs,1)
        total+=batch_labels.shape[0]
        correct+=(predictes==batch_labels).sum().item()
print(correct/total)
```
